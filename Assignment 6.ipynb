```python
import pandas as pd
df = pd.read_csv("C:/Users/ghare/Downloads/Titanic-Dataset.csv")
print(df)
```

```python
q1 = df[(df["Sex"] == "female") &
        (df["Survived"] == 1) &
        (df["Age"] > 30)][["Name", "Age", "Fare", "Pclass"]]

print(q1)
```

```python
q2 = df.nlargest(10, "Fare")[["Name", "Fare", "Pclass"]]
print(q2)
```

```python
q3 = (df.groupby("Pclass")["Survived"].mean() * 100)
q3 = q3.sort_values(ascending=False)
print(q3)
```

```python
q4 = df.groupby("Sex")[["Age", "Fare"]].mean()
print(q4)
```

```python
q5 = df[(df["SibSp"] >= 2) & (df["Parch"] >= 1)]
print(q5)
```

```python
q6 = df.groupby(["Sex", "Pclass"]).size()
print(q6)
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
q8 = df.groupby(["Sex", "Embarked"])["Survived"].mean() * 100
print(q8)
```

```python
average_fare = df["Fare"].mean()

q9 = df[df["Fare"] > average_fare][["Name", "Fare", "Pclass"]]
print(q9)
```

```python
q10 = df.groupby("Pclass").agg(
    TotalPassengers=("PassengerId", "count"),
    Survivors=("Survived", "sum"),
    SurvivalPercentage=("Survived", lambda x: x.mean() * 100)
)

print(q10.round(2))
```
