```python
import matplotlib.pyplot as plt
counts = df["Pclass"].value_counts().sort_index()
plt.bar(counts.index, counts.values)
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Number of Passengers in Each Class")
plt.xticks([1, 2, 3])

plt.show()
```

```python
counts = df["Survived"].value_counts().sort_index()
plt.bar(counts.index, counts.values, color ="hotpink", width = 0.5)
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Titanic Passenger Survival")
plt.xticks([0, 1], ["Not Survived", "Survived"])
plt.show()
```

```python
counts = df.groupby(["Survived", "Sex"]).size().unstack()

print(counts)

counts.plot(kind="bar")

plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Titanic Survival by Gender")

plt.xticks([0, 1], ["Not Survived", "Survived"], rotation=0)

plt.show()
```

```python
data = df.dropna(subset=["Age", "Fare"])
plt.scatter(data["Age"], data["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare in Titanic Dataset")
plt.show()
```

```python
data = df.dropna(subset=["Pclass", "Fare"])
plt.scatter(data["Pclass"], data["Fare"], color ="hotpink")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Passenger Class vs Fare")
plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

```python
data = df.dropna(subset=["Pclass", "Age"])
plt.scatter(data["Pclass"], data["Age"])

plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.title("Passenger Class vs Age")

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

```python
avg_fare = df.groupby("Pclass")["Fare"].mean()

plt.plot(avg_fare.index, avg_fare.values, marker="*", color = "hotpink")

plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.title("Average Fare by Passenger Class")

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

```python
import pandas as pd

data = {
    "Student": ["Amit", "Bhavna", "Chetan", "Diya", "Esha", "Farhan"],
    "Maths": [75, 82, 68, 90, 72, 85],
    "Python": [80, 78, 70, 92, 75, 88],
    "Attendance": [85, 90, 78, 95, 80, 92],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)

df.to_csv("sample_data_assignment_07.csv", index=False)

print("CSV file created successfully!")
print(df)
```

```python
plt.figure(figsize=(7, 4.5))
plt.hist(df["Python"], bins=5, edgecolor="black")
plt.title("Distribution of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y")
plt.show()
```

```python
city_count = df["City"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(city_count,
        labels=city_count.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Students from Each City")
plt.legend(city_count.index, title="Cities")
plt.show()
```

```python
print("Average Maths Marks:", df["Maths"].mean())

print("Average Python Marks:", df["Python"].mean())

print("Average Attendance:", df["Attendance"].mean())
```
