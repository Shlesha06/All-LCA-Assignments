```python
student_dict = {
    101: {"Name": "Alice", "Branch": "CS", "Marks": (85, 90, 88)},
    102: {"Name": "Bob", "Branch": "IT", "Marks": (75, 80, 85)},
}

roll_list = [101, 102]

print("Initial Dictionary:", student_dict)

new_roll = 103
student_dict[new_roll] = {"Name": "Charlie", "Branch": "EC", "Marks": (90, 92, 89)}
roll_list.append(new_roll)

student_dict[102]["Branch"] = "Data Science"
student_dict[102]["Marks"] = (80, 85, 90)

del student_dict[101]
roll_list.remove(101)

print("\n--- Final Student Records ---")
for roll in roll_list:
    name = student_dict[roll]["Name"]
    branch = student_dict[roll]["Branch"]
    marks = student_dict[roll]["Marks"]

    print(
        f"Roll No: {roll} | Name: {name} | Branch: {branch} | Marks (Tuple): {marks}"
    )
```
