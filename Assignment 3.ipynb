```python
def triangle(side1, side2, side3):
    sides = [side1, side2, side3]
    sides.sort()

    a = sides[0]
    b = sides[1]
    c = sides[2]  

    if (a**2) + (b**2) == (c**2):
        return True
    else:
        return False

s1 = float(input("Enter length of side 1: "))
s2 = float(input("Enter length of side 2: "))
s3 = float(input("Enter length of side 3: "))

if triangle(s1, s2, s3):
    print("Yes, it is a right-angled triangle.")
else:
    print("No, it is not a right-angled triangle.")
```
