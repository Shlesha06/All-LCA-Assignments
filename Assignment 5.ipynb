```python
pan = input("Enter PAN Number: ")

if len(pan) != 10:
    print("Invalid PAN Number")
else:
    is_valid = True 
    
    for i in range(0, 5):
        if not (pan[i] >= 'A' and pan[i] <= 'Z'):
            is_valid = False

    for i in range(5, 9):
        if not (pan[i] >= '0' and pan[i] <= '9'):
            is_valid = False

    if not (pan[9] >= 'A' and pan[9] <= 'Z'):
        is_valid = False

    if is_valid:
        print("Valid PAN Number")
    else:
        print("Invalid PAN Number")
```
